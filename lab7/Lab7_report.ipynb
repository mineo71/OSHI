Data Analysis

1. Distribution of Hourly Ambulance Calls

	•	The cnt distribution is right-skewed, with most calls concentrated at lower values.
	•	A few hours exhibit high call volumes, indicating potential outliers or special events.

2. Seasonal Trends

	•	Call volumes vary by season, with peaks in specific periods.
	•	Summer and fall showed higher median call volumes, possibly due to increased motorbike activity.

3. Hourly Patterns

	•	Calls rise sharply during morning and evening rush hours (8-9 AM, 5-6 PM).
	•	Low call volumes are observed at night (12-5 AM).

4. Weather Impact

	•	Severe weather (categories 3 and 4) correlates with higher call volumes.
	•	Clear and misty conditions (categories 1 and 2) have lower call volumes.

5. Correlation Analysis

	•	temp and atemp showed strong multicollinearity (correlation ~0.99).
	•	cnt exhibited moderate correlation with temp, hum, and weathersit.
	•	Weak correlation was observed for windspeed.

Regression Model

Modeling Approach

• Gradient Boosting Regressor was chosen for its robustness and interpretability.
• Features used: season, hr, temp, atemp, hum, weathersit, weekday, workingday, etc.

Evaluation Metrics

• RMSE: Measures prediction error magnitude.
• R² Score: Assesses model fit quality.

Model Performance

• The model achieved a reasonable RMSE and R², suggesting a good fit for predicting cnt.

Feature Importance

• The most influential predictors were hr, temp, and weathersit, reflecting the significant impact of time and weather on call volumes.

Event and Anomaly Detection

Anomalies were detected using Isolation Forest, a robust method for identifying outliers in high-dimensional data.

Findings

• Anomalies corresponded to unusual spikes in ambulance calls.
• Potential causes include major accidents, severe weather events, or public holidays.

Visual Confirmation

• Scatter plots highlighted these anomalies, providing a clearer understanding of unexpected events.

Conclusions

1. Hourly Trends:
• Morning and evening rush hours are critical times for ambulance services.
• Deployment strategies could prioritize these periods for faster response.
	
2. Weather Impact:
• Severe weather increases call volumes significantly.
• Public awareness campaigns could focus on safe riding during adverse weather.
	
3. Seasonal Variations:
• Higher activity in summer and fall aligns with increased outdoor activity and travel.
	
4. Anomaly Detection:
• Isolated high-call hours could indicate infrastructure issues or public safety events.
• Regular monitoring can help in resource allocation.

Recommendations:

•	Real-time weather inputs could improve anomaly detection accuracy.
